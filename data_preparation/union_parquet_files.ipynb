```python
import pyarrow.dataset as ds
```

```python
UNION_DIR = "/home/pi-vault/projects/bilbasen_webscraping/data/union"

def load_until_external_id(path: str):
    dataset = ds.dataset(path, format="parquet")
    names = dataset.schema.names
    cols = names[: names.index("externalId") + 1]
    return dataset.to_table(columns=cols).to_pandas()

paths = {
    "El": "/home/pi-vault/projects/bilbasen_webscraping/data/El",
    "Benzin": "/home/pi-vault/projects/bilbasen_webscraping/data/Benzin",
    "Diesel": "/home/pi-vault/projects/bilbasen_webscraping/data/Diesel",
}

dfs = {k: load_until_external_id(p) for k, p in paths.items()}

for k, df in dfs.items():
    df.to_parquet(f"{UNION_DIR}/{k}_union.parquet", index=False, compression="zstd")
    print("{} rows: {}, unique externalId's: {}".format(k, len(df), df["externalId"].nunique(dropna=False)))
```
